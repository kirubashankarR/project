```python
# Install dependencies
!pip install -q ultralytics opencv-python

import cv2
from ultralytics import YOLO
from collections import defaultdict
import os

# Load YOLOv8 model
model = YOLO("yolov8n.pt")

# Load video
video_path = "/content/853908-hd_1920_1080_25fps.mp4"
cap = cv2.VideoCapture(video_path)

# Video properties
fps = int(cap.get(cv2.CAP_PROP_FPS))
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Output video setup
output_path = "perpendicular_lane_output.mp4"
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

# Define horizontal (perpendicular) lane lines
lane_lines = [height // 3, (height // 3) * 2]  # You can add more lines
lane_counts = [0 for _ in lane_lines]

# COCO vehicle classes
vehicle_classes = {
    2: "car",
    3: "motorcycle",
    5: "bus",
    7: "truck"
}

counted_ids = [set() for _ in lane_lines]

frame_id = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    frame_id += 1

    results = model(frame)[0]

    for box in results.boxes:
        cls_id = int(box.cls[0])
        conf = float(box.conf[0])
        if cls_id not in vehicle_classes or conf < 0.5:
            continue

        x1, y1, x2, y2 = map(int, box.xyxy[0])
        cx = int((x1 + x2) / 2)
        cy = int((y1 + y2) / 2)
        vehicle_type = vehicle_classes[cls_id]
        vehicle_id = f"{vehicle_type}_{x1}_{y1}"

        # Draw box and label
        cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
        cv2.putText(frame, vehicle_type, (x1, y1 - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 1)
        cv2.circle(frame, (cx, cy), 5, (0, 0, 255), -1)

        # Check crossing of horizontal lanes
        for i, line_y in enumerate(lane_lines):
            if abs(cy - line_y) < 10 and vehicle_id not in counted_ids[i]:  # 10px range
                lane_counts[i] += 1
                counted_ids[i].add(vehicle_id)

    # Draw horizontal lane lines and count display
    for i, line_y in enumerate(lane_lines):
        cv2.line(frame, (0, line_y), (width, line_y), (255, 0, 0), 2)
        cv2.putText(frame, f"Lane {i+1} Count: {lane_counts[i]}",
                    (10, line_y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6,
                    (0, 255, 255), 2)

    out.write(frame)

cap.release()
out.release()
```
